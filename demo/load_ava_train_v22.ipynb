- video id (0)
- middle_frame_timestamp (1)
  - 타임스탬프(초 단위)
  - 해당 시점의 **중앙 프레임**이 라벨링된 기준
- person_box_x1, y1, x2, y2 : (2~5)
  - 사람 바운딩 박스 좌표
  - **정규화된 값(0~1 사이)** -> 원본 프레임 크기와 무관
- action_id (6)
  - 해당 인물에 라벨링된 행동 클래스 ID
  - AVA V2.2에서는 **80개 액션 클래스** 존재
  - 동일한 bbox에 다른 action_id가 할당됨.
- **person_id** (7) : <br>
  박스 고유 식별자 (이 프레임 내에서 같은 인물을 구분하기 위한 ID)

**라벨 ID 매핑**

- AVA dataset은 별도의 **label 파일 (ava_action_list_v2.2.pbtxt)** 로 ID ↔ Action name 매핑이 제공됩니다.

- 예:

    - 1 → bend/bow(굽다/허리를 숙이다) 
    - 12 → "stand"
    - 17 → "talk to"
    - 79 → "listen to person"
    - 80 → work on a commputer (컴퓨터 작업하다)
